````python
def slice_string(s, a, b, c, d):
    # Python slicing is exclusive at the stop index,
    # so add +1 to include indices b and d.
    first_slice = s[a : b + 1]
    second_slice = s[c : d + 1]

    return f"{first_slice} {second_slice}"


# Sample Dataset
s = "HumptyDumptysatonawallHumptyDumptyhadagreatfallAlltheKingshorsesandalltheKingsmenCouldntputHumptyDumptyinhisplaceagain."
a, b, c, d = 22, 27, 97, 102

# Output: Humpty Dumpty
print(slice_string(s, a, b, c, d))
````

````python
def sum_odd_integers_fast(a, b):
    # Ensure 'a' starts on an odd number
    if a % 2 == 0:
        a += 1

    # Step by 2 to jump directly between odd numbers
    return sum(range(a, b + 1, 2))


# Example usage:
a = 4886
b = 9317

print(sum_odd_integers_fast(a, b))
````
