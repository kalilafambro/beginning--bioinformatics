Ros #11

```python
from collections import defaultdict

def frequent_words(text, k):
    counts = defaultdict(int)

    # Slide a window of length k across text to count each k-mer
    for i in range(len(text) - k + 1):
        kmer = text[i:i+k]
        counts[kmer] += 1

    # Find the maximum frequency among all k-mers
    max_count = max(counts.values())

    # Collect all k-mers that achieve the maximum frequency
    most_frequent = [kmer for kmer, count in counts.items() if count == max_count]

    return " ".join(most_frequent)


# Sample Dataset
text = "ACGTTGCATGTCGCATGATGCATGAGAGCT"
k = 4

print(frequent_words(text, k))
```

```python
from collections import defaultdict


def find_clumps(genome, k, L, t):
    clump_patterns = set()
    kmer_counts = defaultdict(int)

    # 1. Initialize frequency count for the very first window of length L
    first_window = genome[:L]
    for i in range(L - k + 1):
        kmer = first_window[i : i + k]
        kmer_counts[kmer] += 1
        if kmer_counts[kmer] >= t:
            clump_patterns.add(kmer)

    # 2. Slide the window of length L across the rest of the genome
    for i in range(1, len(genome) - L + 1):
        # Remove the k-mer sliding out on the left
        outgoing_kmer = genome[i - 1 : i - 1 + k]
        kmer_counts[outgoing_kmer] -= 1

        # Add the k-mer sliding in on the right
        incoming_kmer = genome[i + L - k : i + L]
        kmer_counts[incoming_kmer] += 1

        # Check if the newly added k-mer meets or exceeds threshold t
        if kmer_counts[incoming_kmer] >= t:
            clump_patterns.add(incoming_kmer)

    return " ".join(clump_patterns)


# Sample Dataset
genome = "CGGACTCGACAGATGTGAAGAAATGTGAAGACTGAGTGAAGAGAAGAGGAAACACGACACGACATTGCGACATAATGTACGAATGTAATGTGCCTATGGC"
k = 5
L = 75
t = 4

print(find_clumps(genome, k, L, t))
```

```python
def minimum_skew(genome):
    current_skew = 0
    min_skew = 0
    min_positions = [0]  # Skew(Prefix0) is always 0

    for i, base in enumerate(genome, start=1):
        if base == "G":
            current_skew += 1
        elif base == "C":
            current_skew -= 1

        if current_skew < min_skew:
            min_skew = current_skew
            min_positions = [i]
        elif current_skew == min_skew:
            min_positions.append(i)

    return " ".join(map(str, min_positions))


# Sample Dataset
genome = "CCTATCGGTGGATTAGCATGTCCCTGTACGTTTCGCCGCGAACTAGTTCACACGGCTTGATGGCAAATGGTTTTTCCGGCGACCGTAATCGTCCACCGAG"

print(minimum_skew(genome))
```

```python
def neighbors(pattern, d):
    # Base cases
    if d == 0:
        return {pattern}
    if len(pattern) == 1:
        return {"A", "C", "G", "T"}

    neighborhood = set()
    suffix_neighbors = neighbors(pattern[1:], d)

    for suffix in suffix_neighbors:
        # Check Hamming distance of suffix to pattern's suffix
        suffix_dist = sum(1 for a, b in zip(suffix, pattern[1:]) if a != b)

        if suffix_dist < d:
            # Distance allows changing the first base to any nucleotide
            for base in ["A", "C", "G", "T"]:
                neighborhood.add(base + suffix)
        else:
            # Distance limit reached; first base must match the pattern's first base
            neighborhood.add(pattern[0] + suffix)

    return neighborhood


# Sample Dataset
pattern = "ACG"
d = 1

# Print each neighbor on a new line
result = neighbors(pattern, d)
print("\n".join(result))
```

```python
def hamming_distance(p, q):
    """Computes the number of mismatches between two equal-length strings."""
    return sum(1 for a, b in zip(p, q) if a != b)


def approximate_pattern_matching(pattern, text, d):
    positions = []
    k = len(pattern)

    # Slide a window across text
    for i in range(len(text) - k + 1):
        window = text[i : i + k]
        # Check if mismatches are within allowed threshold d
        if hamming_distance(pattern, window) <= d:
            positions.append(i)

    return " ".join(map(str, positions))


# Sample Dataset
pattern = "ATTCTGGA"
text = "CGCCCGAATCCAGAACGCATTCCCATATTTCGGGACCACTGGCCTCCACGGTACGGACGTCAATCAAATGCCTAGCGGCTTGTGGTTTCTCCTACGCTCC"
d = 3

print(approximate_pattern_matching(pattern, text, d))
```

```python
from collections import defaultdict


def neighbors(pattern, d):
    """Generates all k-mers within Hamming distance d of pattern."""
    if d == 0:
        return {pattern}
    if len(pattern) == 1:
        return {"A", "C", "G", "T"}

    neighborhood = set()
    suffix_neighbors = neighbors(pattern[1:], d)

    for suffix in suffix_neighbors:
        suffix_dist = sum(1 for a, b in zip(suffix, pattern[1:]) if a != b)
        if suffix_dist < d:
            for base in ["A", "C", "G", "T"]:
                neighborhood.add(base + suffix)
        else:
            neighborhood.add(pattern[0] + suffix)

    return neighborhood


def frequent_words_with_mismatches(text, k, d):
    kmer_counts = defaultdict(int)

    # 1. For every window in text, generate its d-neighborhood and count occurrences
    for i in range(len(text) - k + 1):
        window = text[i : i + k]
        window_neighbors = neighbors(window, d)

        for nbr in window_neighbors:
            kmer_counts[nbr] += 1

    # 2. Find maximum frequency achieved
    max_count = max(kmer_counts.values())

    # 3. Collect all k-mers achieving the maximum frequency
    most_frequent = [
        kmer for kmer, count in kmer_counts.items() if count == max_count
    ]

    return " ".join(most_frequent)


# Sample Dataset
text = "ACGTTGCATGTCGCATGATGCATGAGAGCT"
k = 4
d = 1

print(frequent_words_with_mismatches(text, k, d))
```

```python
from itertools import product

# Helper function to generate reverse complement of a DNA string
def reverse_complement(dna):
    complement = {'A': 'T', 'C': 'G', 'G': 'C', 'T': 'A'}
    return "".join(complement[base] for base in reversed(dna))

# Helper function to compute Hamming distance between two equal-length strings
def hamming_distance(p, q):
    return sum(1 for a, b in zip(p, q) if a != b)

def frequent_words_with_mismatches_and_rc(text, k, d):
    # Dictionary to store frequencies for all possible k-mers
    kmer_counts = {}

    # 1. Generate all possible k-mers of length k using alphabet 'A', 'C', 'G', 'T'
    all_kmers = ["".join(p) for p in product('ACGT', repeat=k)]

    # Pre-calculate windows of length k from the text
    text_windows = [text[i:i+k] for i in range(len(text) - k + 1)]

    # 2. For each possible k-mer, count occurrences (including mismatches & reverse complement)
    for pattern in all_kmers:
        rc_pattern = reverse_complement(pattern)
        count = 0

        for window in text_windows:
            # Check pattern match within d mismatches
            if hamming_distance(pattern, window) <= d:
                count += 1
            # Check reverse complement match within d mismatches
            if hamming_distance(rc_pattern, window) <= d:
                count += 1

        kmer_counts[pattern] = count

    # 3. Find maximum count achieved
    max_count = max(kmer_counts.values())

    # 4. Collect all k-mers achieving the maximum count
    frequent_kmers = [kmer for kmer, count in kmer_counts.items() if count == max_count]

    return " ".join(frequent_kmers)


# Sample Dataset
text = "ACGTTGCATGTCGCATGATGCATGAGAGCT"
k = 4
d = 1

print(frequent_words_with_mismatches_and_rc(text, k, d))
```
